# Задача 7: Сдвиг частоты от ионного облака

**Условие:**
Сгусток длиной $L$ производит вторичные заряженные частицы (например, вследствие ионизации остаточного газа). Предполагая, что локальная скорость их образования пропорциональна плотности сгустка, а линейная плотность сгустка постоянна, рассчитайте распределение линейного сдвига бетатронной частоты вдоль сгустка.

**Физическая модель:**

1.  **Образование ионов:**
    Пусть сгусток движется со скоростью $c$ (релятивистский).
    Линейная плотность сгустка $\lambda_{beam}(z)$ постоянна в пределах длины $L$:
    $$ \lambda_{beam}(z) = \lambda_0, \quad 0 \le z \le L $$
    (в системе отсчета сгустка). Или, если рассматривать пролет через фиксированную точку, плотность постоянна во времени пролета $\tau = L/c$.

    Ионы образуются неподвижными (тепловые скорости малы).
    Скорость образования ионов в точке $z$ (вдоль сгустка) пропорциональна количеству протонов, которые уже прошли через эту точку сечения вакуумной камеры.
    
    Рассмотрим точку $s$ внутри сгустка ($0$ — "голова", $L$ — "хвост").
    Количество ионов, накопленных в поперечном сечении к моменту прихода координаты $s$ сгустка, пропорционально времени, в течение которого через это сечение проходил пучок, то есть пропорционально расстоянию от головы сгустка $s$.
    
    Следовательно, линейная плотность ионного облака $\lambda_{ion}(s)$ внутри сгустка растет линейно от головы к хвосту:
    $$ \lambda_{ion}(s) = k \cdot s $$
    где $k$ — коэффициент пропорциональности (зависит от сечения ионизации, давления газа и плотности пучка).

2.  **Сила взаимодействия:**
    Сдвиг бетатронной частоты $\Delta \nu$ пропорционален градиенту силы, действующей на частицу пучка со стороны ионного облака.
    Для частицы пучка, находящейся на позиции $s$ от головы, фокусирующая сила создается плотностью ионов $\lambda_{ion}(s)$ в этом же сечении (так как ионы не двигаются продольно).
    
    Сила со стороны цилиндрического заряда плотностью $\lambda_{ion}$:
    $$ F \propto \lambda_{ion} $$
    Градиент силы (фокусировка) $K \propto \lambda_{ion}$.

3.  **Распределение сдвига частоты:**
    Так как $\lambda_{ion}(s) \propto s$, то и сдвиг частоты (tune shift) для частицы, находящейся на расстоянии $s$ от головы сгустка, будет пропорционален $s$.
    
    $$ \Delta \nu(s) \propto s $$
    
    *   В голове сгустка ($s=0$): ионов еще нет, $\Delta \nu = 0$.
    *   В хвосте сгустка ($s=L$): накоплено максимальное количество ионов, $\Delta \nu = \Delta \nu_{max}$.

**Вывод:**
Распределение линейного сдвига бетатронной частоты вдоль сгустка является **линейным**: от нуля в голове до максимума в хвосте.

Это приводит к разбросу частот (tune spread) внутри сгустка, что может вызывать затухание Ландау, но также и резонансные неустойчивости.

```python
import numpy as np
import matplotlib.pyplot as plt

# Настройка графиков
%matplotlib inline
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12

# Параметры
L = 1.0 # Длина сгустка (условные единицы)
s = np.linspace(0, L, 100)

# Линейная плотность ионов (накопление)
lambda_ion = s / L

# Сдвиг частоты (пропорционален плотности ионов)
delta_nu = lambda_ion 

# Визуализация
plt.figure(figsize=(10, 6))
plt.plot(s, delta_nu, linewidth=2)
plt.fill_between(s, delta_nu, alpha=0.3)

plt.title('Распределение сдвига бетатронной частоты вдоль сгустка')
plt.xlabel('Позиция вдоль сгустка $s$ (0 - голова, L - хвост)')
plt.ylabel('Относительный сдвиг частоты $\Delta \\nu(s) / \Delta \\nu_{max}$')
plt.grid(True)
plt.xlim(0, L)
plt.ylim(0, 1.1)

plt.text(0.1, 0.1, 'Голова сгустка\n(нет ионов)', fontsize=12)
plt.text(0.7, 0.9, 'Хвост сгустка\n(максимум ионов)', fontsize=12)

plt.show()
```
