# Сдвиг бетатронной частоты от ионного облака вдоль сгустка

## Условие задачи

Сгусток длиной $L$ производит вторичные заряженные частицы, например ионы остаточного газа.
Локальная скорость их образования пропорциональна плотности сгустка, а линейная плотность сгустка постоянна.
Требуется рассчитать распределение линейного сдвига бетатронной частоты вдоль сгустка.

## Физическая модель

Сгусток движется с релятивистской скоростью, близкой к $c$, а его линейная плотность постоянна в пределах длины $L$:
$$ \lambda_{beam}(z) = \lambda_0, \qquad 0 \le z \le L. $$
При пролёте через фиксированное сечение вакуумной камеры плотность пучка в этом сечении постоянна в течение времени $\tau = L / c$.
Тепловые скорости ионов малы, и за время пролёта сгустка они не успевают сместиться, поэтому ионы считаются неподвижными.

Пусть координата $s$ отсчитывается от головы сгустка ($s = 0$) к хвосту ($s = L$).
Количество ионов, накопленных в сечении к моменту прихода точки $s$ сгустка, пропорционально времени, в течение которого через это сечение уже прошёл пучок, а значит, расстоянию $s$ от головы.
Линейная плотность ионного облака внутри сгустка растёт линейно от головы к хвосту:
$$ \lambda_{ion}(s) = k s, $$
где коэффициент $k$ определяется сечением ионизации, давлением остаточного газа и плотностью пучка.

Сдвиг бетатронной частоты пропорционален градиенту силы, действующей на частицу пучка со стороны ионного облака.
Ионы не смещаются вдоль оси, и на частицу в точке $s$ действует ионная плотность того же сечения $\lambda_{ion}(s)$.
Сила со стороны цилиндрического заряда с линейной плотностью $\lambda_{ion}$ и её градиент пропорциональны этой плотности, $K \propto \lambda_{ion}$.
Следовательно, сдвиг частоты для частицы на расстоянии $s$ от головы пропорционален $s$:
$$ \Delta \nu(s) \propto s. $$
В голове сгустка ионов ещё нет и $\Delta \nu = 0$, в хвосте накоплено максимальное количество ионов и $\Delta \nu = \Delta \nu_{max}$.

## Численный расчёт

Построено распределение относительного сдвига частоты вдоль сгустка.

```python
import numpy as np
import matplotlib.pyplot as plt

# Настройка графиков
%matplotlib inline
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12

# Параметры
L = 1.0 # Длина сгустка (условные единицы)
s = np.linspace(0, L, 100)

# Линейная плотность ионов (накопление)
lambda_ion = s / L

# Сдвиг частоты (пропорционален плотности ионов)
delta_nu = lambda_ion 

# Визуализация
plt.figure(figsize=(10, 6))
plt.plot(s, delta_nu, linewidth=2)
plt.fill_between(s, delta_nu, alpha=0.3)

plt.title('Распределение сдвига бетатронной частоты вдоль сгустка')
plt.xlabel('Координата вдоль сгустка $s$ (0 соответствует голове, $L$ хвосту)')
plt.ylabel(r'Относительный сдвиг частоты $\Delta \nu(s) / \Delta \nu_{max}$')
plt.grid(True)
plt.xlim(0, L)
plt.ylim(0, 1.1)

plt.text(0.1, 0.1, 'Голова сгустка\n(нет ионов)', fontsize=12)
plt.text(0.7, 0.9, 'Хвост сгустка\n(максимум ионов)', fontsize=12)

plt.show()
```

## Итоги

Распределение линейного сдвига бетатронной частоты вдоль сгустка линейно: от нуля в голове до максимума в хвосте.
Возникающий разброс частот внутри сгустка приводит к затуханию Ландау, но может вызывать и резонансные неустойчивости, поскольку хвост сгустка колеблется на частоте, отличной от частоты головы.
